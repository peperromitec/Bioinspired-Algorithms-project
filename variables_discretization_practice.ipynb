## 1. Two decisions and one fitness value

We choose a position $(x,y)$: $x$ represents the horizontal decision and $y$ the vertical decision. We measure closeness to the target through squared distance:

$$F(x,y)=(x-1.3)^2+(y-2.7)^2.$$

Lower fitness represents greater closeness. In the continuous version, the target belongs to the domain and has zero fitness: it is the minimizer because the sum of squares is nonnegative.

In the discrete representation, we use step $0.5$ and integer genes $k_1,k_2\in\{0,1,\ldots,8\}$:

$$x=0.5k_1,\qquad y=0.5k_2.$$

Each gene encodes one decision. The chromosome contains both genes; an individual represents a candidate and a population collects several individuals. The grid permits $9\times9=81$ chromosomes.
